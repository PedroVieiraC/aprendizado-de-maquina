```python
# =====================================================================
# NOTEBOOK 2: CLASSIFICAÇÃO BINÁRIA COM O DATASET CATS VS DOGS (TAREFA 8)
# =====================================================================

# 1. Importação de Bibliotecas Específicas
import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from sklearn.metrics import confusion_matrix, classification_report

print("Bibliotecas carregadas com sucesso! Versão do TensorFlow:", tf.__version__)

# 2. Carregar o dataset Cats vs Dogs (Código fornecido pelo professor)
print("\nBaixando e carregando o dataset Cats vs Dogs (pode levar alguns minutos)...")
(ds_train, ds_test), ds_info = tfds.load(
    'cats_vs_dogs',                       # Nome do dataset no TFDS
    split=['train[:80%]', 'train[80%:]'], # 80% para treino e 20% para teste
    with_info=True,                       # Retorna metadados
    as_supervised=True                    # Formato (imagem, label)
)

# 3. Função de Pré-processamento (Código fornecido pelo professor)
def preprocess(image, label):
    image = tf.image.resize(image, (128, 128)) # Redimensiona para 128x128 pixels
    image = tf.cast(image, tf.float32) / 255.0 # Normaliza os pixels entre [0,1]
    return image, label

# Aplica as transformações, define lotes (batches) de 32 e otimiza o carregamento
ds_train = ds_train.map(preprocess).batch(32).prefetch(1)
ds_test = ds_test.map(preprocess).batch(32).prefetch(1)
print("Dados carregados e pré-processados!")

# 4. Visualizar uma amostra dos dados para conferência
classes_binary = ['Gato', 'Cachorro']
plt.figure(figsize=(10, 4))
for images, labels in ds_train.take(1):
    for i in range(5):
        plt.subplot(1, 5, i + 1)
        plt.imshow(images[i].numpy())
        plt.title(classes_binary[int(labels[i].numpy())])
        plt.axis('off')
plt.tight_layout()
plt.show()

# 5. Criação da CNN Robusta (Baseada no aprendizado das tarefas anteriores)
# Nota: Imagens 128x128 exigem um pouco mais de profundidade que as de 32x32 do CIFAR
model_binary = Sequential([
    # Bloco Convolucional 1
    Conv2D(32, (3,3), activation='relu', padding='same', input_shape=(128, 128, 3)),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    Dropout(0.25),

    # Bloco Convolucional 2
    Conv2D(64, (3,3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    Dropout(0.25),

    # Bloco Convolucional 3
    Conv2D(128, (3,3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2,2)),
    Dropout(0.25),

    # Camadas Densas de Classificação
    Flatten(),
    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.5),

    # Camada de Saída AJUSTADA (Solicitada pelo professor)
    Dense(1, activation='sigmoid')
])

# Compilação utilizando Adam (melhor otimizador da Tarefa 5) e perda Binária
model_binary.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model_binary.summary()

# 6. Treinamento do Modelo
# Treinaremos por 10 épocas, o que já é suficiente para demonstrar convergência na GPU
print("\nIniciando o treinamento no Cats vs Dogs...")
history_binary = model_binary.fit(
    ds_train,
    epochs=10,
    validation_data=ds_test
)

# 7. Plotar curva de loss e acurácia
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(history_binary.history['loss'], label='Treino')
plt.plot(history_binary.history['val_loss'], label='Validação')
plt.title("Loss (Perda Binária)")
plt.legend()

plt.subplot(1,2,2)
plt.plot(history_binary.history['accuracy'], label='Treino')
plt.plot(history_binary.history['val_accuracy'], label='Validação')
plt.title("Acurácia Binária")
plt.legend()
plt.show()

# 🧪 8. Avaliação Final e Métricas Avançadas
print("\nGerando Matriz de Confusão e Relatório...")

# Extrair labels reais e previsões do gerador do TFDS
y_true_bin = []
y_pred_bin = []

for images, labels in ds_test:
    preds = model_binary.predict(images, verbose=0)
    # Como a saída é Sigmoide (probabilidade entre 0 e 1):
    # Se >= 0.5 é classe 1 (Cachorro), senão é classe 0 (Gato)
    pred_classes = (preds >= 0.5).astype(int).flatten()

    y_true_bin.extend(labels.numpy())
    y_pred_bin.extend(pred_classes)

y_true_bin = np.array(y_true_bin)
y_pred_bin = np.array(y_pred_bin)

# Gráfico da Matriz de Confusão
conf_matrix_bin = confusion_matrix(y_true_bin, y_pred_bin)
plt.figure(figsize=(6,5))
sns.heatmap(conf_matrix_bin, annot=True, fmt='d', cmap='Greens',
            xticklabels=classes_binary, yticklabels=classes_binary)
plt.xlabel("Previsto")
plt.ylabel("Verdadeiro")
plt.title("Matriz de Confusão - Cats vs Dogs")
plt.show()

# Relatório de Classificação por Texto
print(classification_report(y_true_bin, y_pred_bin, target_names=classes_binary))
```
