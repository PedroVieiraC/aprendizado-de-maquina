# Arquivo IRIS

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Carregar os dados
df = pd.read_csv('iris.csv')
# Verificar as primeiras linhas
print(df.head())
# Separar atributos e classe
X = df.drop(columns=['species']) # substitua 'species' pelo nome correto da coluna da classe
y = df['species']
# Normalização
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Divisão treino/teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

**Esse trecho de código pode ser separado em 3 pontos**

1-) Carregar as bibliotecas


2-) Define as variáveis de análise


3-) Normaliza os dados e divide em treino e teste

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
k_values = range(1, 31)
accuracies = []
for k in k_values:
  knn = KNeighborsClassifier(n_neighbors=k)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  accuracies.append(acc)
plt.plot(k_values, accuracies, marker='o')
plt.xlabel('Número de Vizinhos (k)')
plt.ylabel('Acurácia')
plt.title('Acurácia em função de k')
plt.grid()
plt.xticks(ticks=range(1, 31, 1))
plt.show()
```

**Trecho de código responsável por gerar o gráfico de fit**

1-) É possível perceber que devido à baixa quantidade de dados no csv fonte, quanto maior o número de vizinhos procurados, pior a acurácia



```python
metrics = ['euclidean', 'manhattan']
for metric in metrics:
  knn = KNeighborsClassifier(n_neighbors=5, metric=metric)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  print(f"Distância: {metric} | Acurácia: {acc:.4f}")
```

**Trecho de código responsável por gerar as distâncias euclideanas e de manhattan**

1) Compara as métricas: avalia o KNN usando as distâncias

2) Treino e teste: Treina o KNN para cada métrica e teste para prever outra classe de dados

3) Acurácia:  Imprime a precisão dos dois modelos de distância\

```python
weights_list = ['uniform', 'distance']
for w in weights_list:
  knn = KNeighborsClassifier(n_neighbors=5, metric='euclidean', weights=w)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  print(f"Peso: {w} | Acurácia: {acc:.4f}")
```

# Arquivo evasao

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Carregar os dados
df = pd.read_csv('evasao.csv')
# Verificar as primeiras linhas
print(df.head())
# Separar atributos e classe
X = df.drop(columns=['Evadiu']) # substitua 'species' pelo nome correto da coluna da classe
y = df['Evadiu']
# Normalização
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Divisão treino/teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

**Esse trecho de código pode ser separado em 3 pontos**

1-) Carregar as bibliotecas

2-) Define as variáveis de análise

3-) Normaliza os dados e divide em treino e teste

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
k_values = range(1, 31)
accuracies = []
for k in k_values:
  knn = KNeighborsClassifier(n_neighbors=k)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  accuracies.append(acc)
plt.plot(k_values, accuracies, marker='o')
plt.xlabel('Número de Vizinhos (k)')
plt.ylabel('Acurácia')
plt.title('Acurácia em função de k')
plt.grid()
plt.xticks(ticks=range(1, 31, 1))
plt.show()
```

*Trecho de código responsável por gerar o gráfico de fit**

1-) O fit desse código também se apresentou diferente tanto do primeiro código quanto do segundo, mantendo uma excelente padronização à partir dos 8 vizinhos mais próximos. Dessa forma, se computacionalmente fosse complexo analisar à partir do 10 (que foi o maior) seria plausível utilizar do oitavo pela padronização dos valores de fit

```python
metrics = ['euclidean', 'manhattan']
for metric in metrics:
  knn = KNeighborsClassifier(n_neighbors=9, metric=metric)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  print(f"Distância: {metric} | Acurácia: {acc:.4f}")
```

**Trecho de código responsável por gerar as distâncias euclideanas e de manhattan**

Estando em outro padrão de analise, é possível perceber que por ter um padrão de valores de fit, a acurácia já é bem maior em relação ao segundo código.

```python
weights_list = ['uniform', 'distance']
for w in weights_list:
  knn = KNeighborsClassifier(n_neighbors=9, metric='euclidean', weights=w)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  print(f"Peso: {w} | Acurácia: {acc:.4f}")
```

**Trecho de código responsável por analisar o impâcto do peso no KNN**

O peso é desconsiderável nessa questão porque a diferença entre as acurácias foram mínimas

# Arquivo santander_train

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Carregar os dados
df = pd.read_csv('santander_train.csv')
# Verificar as primeiras linhas
print(df.head())
# Separar atributos e classe
X = df.drop(columns=['target']) # substitua 'species' pelo nome correto da coluna da classe
y = df['target']
# Normalização
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Divisão treino/teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

1-) Carregar as bibliotecas

2-) Define as variáveis de análise

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Carregar os dados
df = pd.read_csv('santander_train.csv')

sample_fraction = 0.1
df_sampled = df.sample(frac=sample_fraction, random_state=42)
print(f"DataFrame original: {len(df)} linhas. DataFrame amostrado: {len(df_sampled)} linhas.")
# -------------------------------------------------------------------

# Verificar as primeiras linhas do DataFrame amostrado
print(df_sampled.head())

# Separar atributos e classe
X = df_sampled.drop(columns=['target']) # substitua 'species' pelo nome correto da coluna da classe
y = df_sampled['target']

# Normalização
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Divisão treino/teste
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

**Mudança na amostragem de dados para utilizar apenas 19.**

Antes de separar X e y, ele amostra uma porcentagem menor do DataFrame (por exemplo, 10% ou 5%). Você pode ajustar essa porcentagem.
Em seguida, continua com a separação de atributos e classe, normalização e divisão treino/teste, mas agora com um conjunto de dados muito menor, o que deve acelerar drasticamente o processamento.

Foi tomada essa decisão porque a base de dados era muito grande e estava acontecendo keyboardInterrupt.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
k_values = range(1, 31)
accuracies = []
for k in k_values:
  knn = KNeighborsClassifier(n_neighbors=k)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  accuracies.append(acc)
plt.plot(k_values, accuracies, marker='o')
plt.xlabel('Número de Vizinhos (k)')
plt.ylabel('Acurácia')
plt.title('Acurácia em função de k (com amostra reduzida)')
plt.grid()
plt.xticks(ticks=range(1, 31, 1))
plt.show()
```

Observa-se que a acurácia se mantém estável em torno de 0.89-0.90, mesmo com o aumento do número de vizinhos (k). Isso pode sugerir que o modelo KNN é robusto para este conjunto de dados, ou que a amostragem reduzida pode ter suavizado variações maiores que seriam observadas com o dataset completo.

```python
metrics = ['euclidean', 'manhattan']
for metric in metrics:
  knn = KNeighborsClassifier(n_neighbors=7, metric=metric)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  print(f"Distância: {metric} | Acurácia: {acc:.4f}")
```
